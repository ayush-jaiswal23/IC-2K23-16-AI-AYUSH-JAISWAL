```python
#using matplotlib for visualization and networkx for graph representation
import networkx as nx
import matplotlib.pyplot as plt

class SemanticNetwork:
    def __init__(self):
        self.graph = nx.DiGraph()

    def add_relation(self, subject, relation, target):
        self.graph.add_edge(subject, target, relation=relation)
```

```python
net = SemanticNetwork()
net.add_relation("Canary", "is_a", "Bird")
net.add_relation("Bird", "is_a", "Animal")
net.add_relation("Animal", "has_property", "breathes")
net.add_relation("Bird", "has_property", "has_wings")
net.add_relation("Bird", "has_property", "can_fly")
net.add_relation("Canary", "has_property", "sings_melodiously")
```

```python
# 2. Setup the Plot
plt.figure(figsize=(10, 7))
plt.title("Semantic Network Visualization", fontsize=14, color="darkslategray")
```

```python
# 3. Compute a layout for node positioning
# spring_layout positions nodes using a force-directed algorithm
pos = nx.spring_layout(net.graph, k=1.5, seed=42)
```

```python
# 4. Draw nodes and node labels
nx.draw_networkx_nodes(
    net.graph, pos, 
    node_size=3000, 
    node_color="lightblue", 
    edgecolors="darkblue"
)
nx.draw_networkx_labels(
    net.graph, pos, 
    font_size=10, 
    font_weight="bold"
)
nx.draw_networkx_edges(net.graph, pos, arrowstyle="->", arrowsize=20, edge_color="gray", width=2)
```

```python
# 5. Draw directed edges (arrows)
nx.draw_networkx_edges(
    net.graph, pos, 
    arrowstyle="->", 
    arrowsize=20, 
    edge_color="gray", 
    width=2
)
```

```python
# 6. Extract and draw edge labels (the relations: is_a, has_property)
edge_labels = nx.get_edge_attributes(net.graph, 'relation')
nx.draw_networkx_edge_labels(
    net.graph, pos, 
    edge_labels=edge_labels, 
    font_size=9, 
    font_color="darkred"
)
```

```python
# 7. Display the plot
plt.axis("off")
plt.tight_layout()
plt.show()
```
